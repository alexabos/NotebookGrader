# Python Basics Assignment
Practice with lists, loops, and functions.

```python
# Part 1: Lists
fruits = ['apple', 'banana', 'cherry', 'date']
print('Fruits:', fruits)
print('First fruit:', fruits[0])
print('Number of fruits:', len(fruits))
```

```python
# Part 2: Loops
print('All fruits in uppercase:')
for fruit in fruits:
    print(fruit.upper())
```

```python
# Part 3: Functions
def find_longest(items):
    """Return the longest string in a list."""
    longest = ''
    for item in items:
        if len(item) > len(longest):
            longest = item
    return longest

result = find_longest(fruits)
print('Longest fruit name:', result)
```

```python
# Bonus: sum of lengths
def total_length(items):
    return sum(len(item) for item in items)

print('Total characters across all fruits:', total_length(fruits))
```
